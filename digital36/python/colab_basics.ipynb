## 4. 입력 폼으로 한 달 점심값 계산기

코드 끝에 `# @param`을 붙이면 오른쪽에 **입력 칸(슬라이더·체크박스)** 이 생깁니다. 코드를 몰라도 값을 바꿀 수 있어요.
✏️ 빈칸을 채워 실행한 뒤, **슬라이더를 움직여** 보세요. 값을 바꾸면 자동으로 다시 계산됩니다.

````python
# @title 🍚 한 달 점심값 계산기 { run: "auto" }
점심값 = 9000  # @param {type:"slider", min:5000, max:15000, step:500}
주당_횟수 = 5  # @param {type:"integer"}
커피도_마심 = True  # @param {type:"boolean"}

한달_점심값 = None   # ✏️ 빈칸: 점심값 × 주당_횟수 × 4 (한 달 = 4주로 계산)

if 한달_점심값 is None:
    print("⏳ ✏️ 줄을 채우고 다시 실행하세요")
else:
    커피값 = 4500 * 주당_횟수 * 4 if 커피도_마심 else 0
    print(f"한 달 점심값: {한달_점심값:,}원")
    print(f"한 달 커피값: {커피값:,}원")
    print(f"합계        : {한달_점심값 + 커피값:,}원")
    확인("4 점심값", 한달_점심값, 점심값 * 주당_횟수 * 4)
````

## 8. 걸음 수 그래프 그리기

### 8-1. 한글 글꼴 설치 (처음 한 번, 약 20초)
코랩 그래프는 기본 글꼴에 한글이 없어 □□□로 깨집니다. 먼저 나눔고딕 글꼴을 설치합니다.

````python
import os, subprocess
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

글꼴 = "/usr/share/fonts/truetype/nanum/NanumGothic.ttf"
if not os.path.exists(글꼴):
    try:
        subprocess.run(["apt-get", "-qq", "install", "-y", "fonts-nanum"], check=False,
                       stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    except FileNotFoundError:
        pass
if os.path.exists(글꼴):
    fm.fontManager.addfont(글꼴)
    plt.rcParams["font.family"] = "NanumGothic"
    print("✅ 한글 글꼴 준비 완료")
else:
    plt.rcParams["font.family"] = "Malgun Gothic"
    print("코랩이 아닌 곳에서 실행 중 — 맑은 고딕을 씁니다")
plt.rcParams["axes.unicode_minus"] = False
````

### 8-2. 일주일 걸음 수 막대그래프
`걸음` 의 숫자 7개를 스마트폰 건강 앱에서 본 **내 걸음 수**로 바꾸고 실행하세요. (없으면 그대로)

````python
요일 = ["월", "화", "수", "목", "금", "토", "일"]
걸음 = [6200, 8100, 4300, 9800, 7200, 12500, 3100]   # ← 내 걸음 수로 바꾸기
목표 = 7000

평균 = sum(걸음) / len(걸음)
색 = ["#10684A" if n >= 목표 else "#C3CFC6" for n in 걸음]

plt.figure(figsize=(7, 3.4))
plt.bar(요일, 걸음, color=색)
plt.axhline(목표, color="#A94D30", linestyle="--", linewidth=1, label=f"목표 {목표:,}보")
plt.title(f"일주일 걸음 수 (평균 {평균:,.0f}보)")
plt.ylabel("걸음")
plt.legend()
plt.show()

print("목표를 넘은 날:", sum(1 for n in 걸음 if n >= 목표), "일")
````

## 🏆 도전 문제 (선택) — 평 ↔ ㎡ 변환기

드롭다운 폼(`# @param ["…", "…"]`)과 `if / else` 를 씁니다. **1평 = 3.3058㎡** 입니다.
✏️ 두 줄을 채우고, **84㎡ → 약 25.41평**이 나오면 성공!

````python
# @title 📐 평 ↔ ㎡ 변환기 { run: "auto" }
값 = 84  # @param {type:"number"}
방향 = "㎡ → 평"  # @param ["평 → ㎡", "㎡ → 평"]

한_평 = 3.3058
if 방향 == "평 → ㎡":
    결과 = None   # ✏️ 빈칸: 값 × 한_평
    단위 = "㎡"
else:
    결과 = None   # ✏️ 빈칸: 값 ÷ 한_평
    단위 = "평"

if 결과 is None:
    print("⏳ ✏️ 줄을 채우고 다시 실행하세요")
else:
    print(f"{값} ({방향}) = {결과:.2f}{단위}")
````
