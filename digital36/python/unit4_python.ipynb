## 2. 읽어서 리스트로 만들기

## 5. "전자책이 늘면 방문자가 줄었나?" — 방향 세어 보기
달마다 두 값이 **같은 방향**(둘 다 증가/둘 다 감소)인지 **반대 방향**인지 셉니다.

````python
같은방향, 반대방향 = 0, 0
for 월, 방문, 전자, 비고 in 변화:
    if 방문 * 전자 > 0:
        같은방향 += 1
        판정 = "같은 방향"
    elif 방문 * 전자 < 0:
        반대방향 += 1
        판정 = "반대 방향"
    else:
        판정 = "변화 없음"
    print(f"{월:>3}  방문자 {'↑' if 방문 > 0 else '↓'}  전자책 {'↑' if 전자 > 0 else '↓'}  → {판정}")

print(f"\n같은 방향 {같은방향}개월 · 반대 방향 {반대방향}개월")
````

👉 반대 방향이 조금 많지만 **여섯 달 자료**로는 규칙이라 말하기 어렵고, 6월의 큰 감소는 **휴관** 때문입니다.
**함께 변했다고 해서 원인은 아닙니다.** (상관 ≠ 인과)

## 6. (선택) 휴관을 반영해 하루 평균으로 비교
6월은 30일 중 7일(8~14일) 휴관 → 운영일 23일, 5월은 31일 모두 운영했다고 가정합니다.

````python
운영일 = {"5월": 31, "6월": 30 - 7}
for 줄 in 기록[-2:]:
    하루평균 = 줄["방문자"] / 운영일[줄["월"]]
    print(f"{줄['월']} 방문자 {줄['방문자']:,}명 ÷ {운영일[줄['월']]}일 = 하루 평균 {하루평균:.1f}명")

감소_월 = (9800 - 15100) / 15100 * 100
감소_일 = (9800 / 23 - 15100 / 31) / (15100 / 31) * 100
print(f"\n월 합계로 보면 {감소_월:.1f}% 감소, 하루 평균으로 보면 {감소_일:.1f}% 감소")
````

## 7. (선택) 그래프 — 두 값을 따로 그리기
단위가 다른 두 값을 한 그래프에 넣으면 작은 값의 변화가 안 보입니다. 두 칸으로 나눠 그립니다.

````python
import os, subprocess
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

글꼴 = "/usr/share/fonts/truetype/nanum/NanumGothic.ttf"
if not os.path.exists(글꼴):
    try:
        subprocess.run(["apt-get", "-qq", "install", "-y", "fonts-nanum"], check=False,
                       stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    except FileNotFoundError:
        pass
if os.path.exists(글꼴):
    fm.fontManager.addfont(글꼴)
    plt.rcParams["font.family"] = "NanumGothic"
else:
    plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

월들 = [줄["월"] for 줄 in 기록]
fig, (위, 아래) = plt.subplots(2, 1, figsize=(7, 5), sharex=True)
위.plot(월들, [줄["방문자"] for 줄 in 기록], marker="o", color="#10684A")
위.set_title("방문자 수 (6월 휴관 7일)")
아래.plot(월들, [줄["전자책"] for 줄 in 기록], marker="o", color="#1F5E86")
아래.set_title("전자책 대출")
plt.suptitle("한빛시 도서관 (수업용 가상 데이터)")
plt.tight_layout()
plt.show()
````
