```python
# 테스트 도우미 — 먼저 실행하세요
def 테스트(함수, 사례):
    통과 = 0
    for 입력, 기대 in 사례:
        try:
            결과 = 함수(입력)
        except Exception as e:
            결과 = f"오류: {type(e).__name__}"
        ok = 결과 == 기대
        통과 += ok
        print(("✅" if ok else "❌"), f"{함수.__name__}({입력!r}) → {결과!r}   (기대: {기대!r})")
    print(f"--- {통과}/{len(사례)} 통과")
    return 통과 == len(사례)

print("준비 완료")
```

### 코드 3 — 요청: "1부터 n까지 더하는 함수를 만들어 줘"
👉 `합계(10)`은 55여야 합니다. 1+2+…+10을 떠올려 보세요.

```python
# 🤖 AI가 준 코드
def 합계(n):
    total = 0
    for i in range(1, n):
        total += i
    return total

print("합계(10) =", 합계(10))
테스트(합계, [(1, 1), (3, 6), (10, 55), (100, 5050)])
```
