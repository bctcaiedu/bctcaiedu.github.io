### 코드 3 — 요청: "점수가 60점 이상이면 합격이라고 알려 줘"
여러 점수로 시험합니다. **딱 60점**인 사람은 어떻게 나올까요?

````python
# 🤖 AI가 준 코드
점수들 = [95, 72, 60, 59, 30]

for 점수 in 점수들:
    if 점수 > 60:
        결과 = "합격"
    else:
        결과 = "불합격"
    print(점수, "점 →", 결과)
````

"60점 **이상**"은 60점도 포함합니다. `>`는 **초과**(60 제외), `>=`는 **이상**(60 포함)입니다.
✏️ 비교 기호 하나를 고치세요.

````python
def 합격인가(점수):
    return 점수 > 60   # ✏️ 이 줄의 기호를 고치세요

for 점수 in [95, 72, 60, 59, 30]:
    print(점수, "점 →", "합격" if 합격인가(점수) else "불합격")
확인("딱 60점", 합격인가(60), True)
확인("59점", 합격인가(59), False)
````

- 코드 3을 95점, 30점으로만 시험했다면 틀린 곳을 찾았을까요? → **경계값(60점)으로 시험하기**

## 도전 A. 윤년 판정 · 1부터 n까지 합
### A-1. 요청: "연도를 넣으면 윤년인지 알려 줘"
**윤년 규칙**: 4로 나누어떨어지면 윤년. 단, 100으로 나누어떨어지면 평년. 단, 400으로 나누어떨어지면 다시 윤년.
AI가 준 코드는 `return 연도 % 4 == 0` 한 줄이었습니다. 1900년, 2100년을 넣으면? 규칙대로 고치세요.

````python
def 윤년인가(연도):
    return 연도 % 4 == 0   # ✏️ 규칙대로 고치기 (if를 여러 번 써도 됩니다)

for 연도, 기대 in [(2024, True), (2023, False), (1900, False), (2000, True), (2100, False)]:
    확인(str(연도) + "년", 윤년인가(연도), 기대)
````

### A-2. 요청: "1부터 n까지 더해 줘"
`합계(10)`은 55여야 합니다. 힌트: `range(1, n)`은 n을 포함할까요? 가우스 공식 `n × (n + 1) ÷ 2`로 교차 확인해 보세요.

````python
def 합계(n):
    total = 0
    for i in range(1, n):   # ✏️ 어디가 틀렸을까?
        total += i
    return total

for n in [1, 3, 10, 100]:
    확인("합계(" + str(n) + ")", 합계(n), n * (n + 1) // 2)
````
