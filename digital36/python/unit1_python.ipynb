## 2. "오늘" 다음에 오는 단어 세기 (5차시)

ChatGPT 같은 생성형 AI는 **"지금까지의 글 다음에 올 단어"** 를 고르면서 문장을 만듭니다.
그 기준은 **학습한 글에서 자주 나온 단어**입니다. 아주 작은 데이터로 직접 확인해 봅니다.

### 2-1. 문장을 단어로 쪼개기

````python
문장들 = [
    "나는 오늘 아침에 커피를 마셨다",
    "나는 오늘 점심에 김밥을 먹었다",
    "나는 오늘 저녁에 친구를 만났다",
    "나는 어제 아침에 운동을 했다",
    "오늘 아침에 비가 왔다",
    "오늘 점심에 회의가 있었다",
    "오늘 아침에 버스를 탔다",
]

단어들 = []
for 문장 in 문장들:
    단어들 = 단어들 + 문장.split()

print(단어들)
print("단어 수:", len(단어들))
````

### 2-2. ✏️ "오늘" 다음 단어 모으기
`단어들[i]`가 찾는 단어이면, **바로 다음 단어** `단어들[i + 1]`을 `다음단어` 리스트에 넣습니다.
리스트에 넣을 때는 `리스트.append(값)`을 씁니다.

````python
찾을단어 = "오늘"
다음단어 = []

for i in range(len(단어들) - 1):
    if 단어들[i] == 찾을단어:
        pass   # ✏️ 빈칸: pass를 지우고 다음단어 리스트에 단어들[i + 1]을 넣으세요

print(찾을단어, "다음에 나온 단어들:", 다음단어)
````

````python
확인('"오늘" 다음 단어 개수', len(다음단어) or None, 6)
확인('"아침에"가 나온 횟수', 다음단어.count("아침에") or None, 3)
````

### 2-3. 몇 번씩 나왔나? — AI의 선택
`리스트.count(값)`은 그 값이 몇 번 들어 있는지 셉니다.

````python
이미_센_단어 = []
가장_많은_단어 = ""
가장_많은_횟수 = 0

for 단어 in 다음단어:
    if 단어 not in 이미_센_단어:
        횟수 = 다음단어.count(단어)
        비율 = 횟수 / len(다음단어) * 100
        print(찾을단어, "→", 단어, ":", 횟수, "번", "■" * 횟수, round(비율), "%")
        이미_센_단어.append(단어)
        if 횟수 > 가장_많은_횟수:
            가장_많은_단어 = 단어
            가장_많은_횟수 = 횟수

print()
print("🤖 AI의 선택:", 찾을단어, 가장_많은_단어)
````

### 3-1. ✏️ 의심 단어 세기
`단어 in 문자`는 문자 안에 그 단어가 들어 있으면 `True`입니다.

````python
의심단어 = ["택배", "주소", "결제", "환급", "http", "클릭", "본인 확인", "계좌"]

문자 = "[택배] 주소 불일치로 보관 중입니다. 확인하세요: http://abc.xyz"

개수 = 0
for 단어 in 의심단어:
    if 단어 in 문자:
        print("  발견:", 단어)
        pass   # ✏️ 빈칸: pass를 지우고 개수를 1 늘리세요

print("의심 단어", 개수, "개")
if 개수 >= 3:
    print("🔴 위험! 링크를 누르지 마세요")
elif 개수 >= 1:
    print("🟡 주의! 보낸 사람을 확인하세요")
else:
    print("🟢 의심 단어 없음 (그래도 모르는 링크는 누르지 않기)")
````

아래 셀을 **여러 번** 실행해 보세요. 확률에 따라 단어를 뽑기 때문에 문장이 매번 달라집니다. (실제 AI도 약간의 무작위성을 둡니다)

````python
import random

def 문장_만들기(표, 시작, 길이=6):
    결과 = [시작]
    지금 = 시작
    for _ in range(길이 - 1):
        if 지금 not in 표:
            break
        후보 = 표[지금]
        지금 = random.choices(list(후보), weights=list(후보.values()))[0]
        결과.append(지금)
    return " ".join(결과)

for _ in range(3):
    print(문장_만들기(표, "나는"))
````

## 도전 B. 비밀번호 생성기 · 강도 검사기
> ⚠️ **지금 실제로 쓰는 비밀번호는 절대 입력하지 마세요.** 예시 문자열로만 시험합니다.

`random`은 다음 값을 예측할 수 있는 방식이라, 비밀번호처럼 보안이 필요한 곳에는 `secrets` 모듈을 씁니다.

````python
import secrets
import string

특수문자 = "!@#$%^&*"

def 비밀번호_만들기(길이=14):
    문자 = string.ascii_letters + string.digits + 특수문자
    while True:
        pw = "".join(secrets.choice(문자) for _ in range(길이))
        if (any(c.islower() for c in pw) and any(c.isupper() for c in pw)
                and any(c.isdigit() for c in pw) and any(c in 특수문자 for c in pw)):
            return pw

for _ in range(3):
    print(비밀번호_만들기())
````

**점수 규칙 (0~5점)**: 흔한 비밀번호면 0점 · 12자 이상 +1 · 16자 이상 +1 · 대·소문자 섞임 +1 · 숫자 +1 · 특수문자 +1 · 같은 문자 3번 연속 −1 · 연속 숫자(123) −1

**TODO 4곳**을 채우세요. 힌트: `any(c.isupper() for c in pw)`, `not c.isalnum()`, `pw[i] == pw[i+1] == pw[i+2]`

````python
흔한_비밀번호 = ["password", "123456", "12345678", "qwerty", "qwer1234",
               "111111", "abc123", "iloveyou", "admin", "1q2w3e4r"]

def 강도_검사(pw):
    if pw.lower() in 흔한_비밀번호:
        return 0, ["누구나 아는 흔한 비밀번호입니다"]
    점수 = 0
    조언 = []
    if len(pw) >= 12:
        점수 += 1
    else:
        조언.append("12자 이상으로 늘리세요")
    if len(pw) >= 16:
        점수 += 1

    # ✏️ TODO 1: 소문자와 대문자가 모두 있으면 +1, 아니면 조언 추가
    # ✏️ TODO 2: 숫자가 있으면 +1, 아니면 조언 추가
    # ✏️ TODO 3: 특수문자(isalnum()이 False인 글자)가 있으면 +1, 아니면 조언 추가

    for i in range(len(pw) - 2):
        # ✏️ TODO 4: 같은 문자가 3번 연속이면 점수 -1, 조언 추가, break
        pass

    for i in range(len(pw) - 2):
        a, b, c = pw[i], pw[i + 1], pw[i + 2]
        if a.isdigit() and b.isdigit() and c.isdigit() and int(b) == int(a) + 1 and int(c) == int(b) + 1:
            점수 -= 1
            조언.append("123처럼 연속된 숫자는 피하세요")
            break
    return max(0, min(점수, 5)), 조언
````

````python
테스트 = [("password", 0), ("abcd1234", 0), ("Summer2026!", 3),
          ("coffee-tree-moon-47", 4), ("Aaaa1111!!", 2)]
통과 = 0
for pw, 기대 in 테스트:
    점수, 조언 = 강도_검사(pw)
    ok = 점수 == 기대
    통과 += ok
    print(("✅" if ok else "❌"), pw, "→", 점수, "점 (기대", 기대, "점)", 조언)
print(통과, "/", len(테스트), "통과")
````
