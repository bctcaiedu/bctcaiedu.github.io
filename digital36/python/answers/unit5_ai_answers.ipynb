### 이 노트북에서 AI를 부르는 방법
파이썬 함수 **`AI("질문")`** 한 줄로 생성형 AI에게 묻고, 답을 **문자열**로 돌려받습니다. 받은 답은 리스트에 담고, `for`로 돌리고, `if`로 검사할 수 있습니다.

| 순서 | 연결 방식 | 필요한 것 |
|---|---|---|
| ① | **Colab 내장 AI** (`google.colab.ai`) | 구글 계정 로그인만. 계정 · 지역에 따라 아직 열리지 않았거나 사용량 제한이 있을 수 있음 |
| ② | **Gemini API 키** | [Google AI Studio](https://aistudio.google.com/apikey)에서 무료 키 발급 → 왼쪽 🔑 **보안 비밀**에 이름 `GOOGLE_API_KEY`로 저장 → 노트북 액세스 켜기 |
| ③ | **연습 모드** | 아무것도 없어도 됨 — 미리 준비한 답으로 파이썬 부분을 끝까지 연습 |

①이 되면 ①, 안 되면 ②, 둘 다 안 되면 ③으로 **자동으로** 넘어갑니다. 아래 셀을 실행하면 지금 어떤 방식인지 알려 줍니다.

> ⚠️ **개인정보 금지** — 이름 · 전화번호 · 주민번호 · 계좌 · 비밀번호는 AI에 보내지 않습니다. AI의 답은 **참고용**이고, 사실은 파이썬과 공식 자료로 다시 확인합니다.

```python
# 🤖 AI 연결 도우미 — 맨 먼저 한 번 실행하세요 (고칠 필요 없음)
import textwrap, random

제미나이_모델 = "gemini-2.5-flash"   # ② API 키 방식에서 쓸 모델 (서비스가 바뀌면 이름만 고치세요)
연습_규칙 = []                        # ③ 연습 모드에서 쓸 (키워드, 답) 목록 — 미션마다 채웁니다
_방식 = None

def _연습_답(질문):
    for 키워드, 답 in 연습_규칙:
        if 키워드 in 질문:
            return 답() if callable(답) else 답
    return random.choice([
        "(연습 모드) 좋은 질문이에요. 실제 AI에 연결되면 이 자리에 진짜 답이 나옵니다.",
        "(연습 모드) 요청하신 내용을 정리하면 다음과 같습니다. 1) 핵심 확인 2) 실행 3) 점검",
        "(연습 모드) 지금은 미리 준비한 답입니다. 결과를 꼭 다시 확인하세요.",
    ])

def AI(질문, 보여주기=False):
    """생성형 AI에게 질문하고 답(문자열)을 돌려줍니다."""
    global _방식
    질문 = str(질문)
    답 = None
    if _방식 in (None, "Colab 내장 AI"):
        try:
            from google.colab import ai as _colab_ai
            답 = str(_colab_ai.generate_text(질문)).strip()
            _방식 = "Colab 내장 AI"
        except Exception:
            답 = None
    if 답 is None and _방식 in (None, "Gemini API 키"):
        try:
            from google.colab import userdata
            from google import genai
            _client = genai.Client(api_key=userdata.get("GOOGLE_API_KEY"))
            답 = _client.models.generate_content(model=제미나이_모델, contents=질문).text.strip()
            _방식 = "Gemini API 키"
        except Exception:
            답 = None
    if 답 is None:
        _방식 = "연습 모드"
        답 = _연습_답(질문)
    if 보여주기:
        for 줄 in 답.split("\n"):
            print(textwrap.fill(줄, 70) if 줄 else "")
    return 답

def 확인(이름, 결과, 정답):
    if 결과 is None:
        print("⏳", 이름, ": 아직 비어 있어요. ✏️ 줄을 채우고 다시 실행하세요")
    elif 결과 == 정답:
        print("✅", 이름, ": 정답입니다!")
    else:
        print("❌", 이름, ": 결과", 결과, "/ 기대", 정답)

시험 = AI("한 문장으로 인사해 줘. 이모지 하나 넣어서.")
print("연결 방식:", _방식)
print("AI ▶", 시험)
if _방식 == "연습 모드":
    print("ℹ️ 실제 AI에 연결되지 않아 연습 모드로 진행합니다. 파이썬 부분은 똑같이 연습할 수 있어요.")
```

## A9. 나만의 AI 비서 지침서 (4차시)

### A9-1. 비서 설계도
값을 내 목적에 맞게 바꾸세요.

```python
비서 = {
    "이름": "걷기 코치 튼튼이",
    "역할": "하루 만 보 걷기 습관을 돕는 코치",
    "대상": "운동을 막 시작한 50대 초보자",
    "말투": "존댓말, 밝고 짧게, 이모지 1개",
    "답 형식": "3줄 이내, 오늘 바로 할 수 있는 행동 1가지를 꼭 포함",
}
금지 = [
    "병을 진단하거나 약을 권하지 않는다",
    "개인정보(이름, 주소, 전화번호)를 묻지 않는다",
    "모르는 것은 모른다고 말한다",
]
print("비서 설계 완료:", 비서["이름"])
```

### A9-2. ✏️ 지침서 자동 조립
딕셔너리는 `for 키, 값 in 비서.items()`로, 금지 목록은 `for 항목 in 금지`로 돌면서 `지침` 문자열에 한 줄씩 붙입니다.

```python
def 지침서_만들기(비서, 금지):
    지침 = "너는 아래 설정을 지키는 AI 비서야.\n"
    for 키, 값 in 비서.items():
        지침 += f"- {키}: {값}\n"
    지침 += "[하지 말 것]\n"
    for 항목 in 금지:
        지침 += f"- {항목}\n"
    return 지침

지침 = 지침서_만들기(비서, 금지)
print(지침)
```

```python
확인("금지 사항이 모두 들어갔나", all(항목 in 지침 for 항목 in 금지) or None, True)
```

### A9-3. 지침 없음 vs 지침 있음
같은 질문을 두 번 보냅니다. 지침이 있을 때 **말투 · 길이 · 형식**이 설계대로 나오는지 보세요. 마음에 들면 위 지침서를 복사해 **Gemini 젬 만들기**의 지침 칸에 붙여 넣습니다.

```python
연습_규칙.append(("설정을 지키는", "오늘 점심 뒤 10분만 걸어 보세요! 🚶 걷기 전 가볍게 발목을 돌리면 좋아요. 무리하지 말고 천천히 시작해요."))
연습_규칙.append(("걷", "걷기는 건강에 좋은 운동입니다. 하루 30분 이상 꾸준히 걷는 것을 권장하며, 바른 자세와 편한 신발이 중요합니다. 무릎이 아프면 전문가와 상담하세요. 또한 수분 섭취도 잊지 마세요."))

시험질문 = ["걷기를 매일 하려면 어떻게 해야 해?", "비 오는 날엔 걷기 대신 뭘 하지?", "무릎이 조금 아픈데 계속 걸어도 돼?"]
for q in 시험질문:
    print("❓", q)
    print("  [지침 없음]", AI(q)[:120])
    print("  [지침 있음]", AI(지침 + "\n질문: " + q)[:120], "\n")
```

## A10. AI 루틴 코치 (5차시)

### A10-1. 내 일주일 기록
월요일부터 일요일까지 7칸입니다. **O = 했음, X = 못 함**.

```python
기록 = {
    "아침 물 한 잔": ["O", "O", "O", "X", "O", "O", "O"],
    "만 보 걷기":   ["O", "X", "O", "X", "X", "O", "O"],
    "자기 전 폰 끄기": ["X", "X", "O", "X", "X", "X", "O"],
    "AI로 영어 한 문장": ["O", "O", "O", "O", "X", "O", "O"],
}
for 습관, 칸 in 기록.items():
    print(f"{습관:<12} {' '.join(칸)}")
```

### A10-2. ✏️ 달성률 계산
**달성률(%) = O 개수 ÷ 7 × 100** 입니다. `칸.count("O")`로 O 개수를 셉니다.

```python
달성률 = {}
for 습관, 칸 in 기록.items():
    률 = round(칸.count("O") / len(칸) * 100)
    달성률[습관] = 률
    막대 = "■" * (칸.count("O")) + "□" * (len(칸) - 칸.count("O"))
    print(f"{습관:<12} {막대} {률}%")
```

```python
확인("아침 물 한 잔 달성률", 달성률.get("아침 물 한 잔"), 86)
```

### A10-3. AI 코치의 응원 보고서
파이썬이 계산한 **숫자만** 보내고, 글은 AI가 씁니다. 결과를 파일로 저장해 루틴북에 붙여 넣으세요.

```python
연습_규칙.insert(0, ("루틴 코치", "📋 이번 주 잘한 점: 아침 물 한 잔을 거의 매일 지켰어요!\n🔧 다음 주 제안: 자기 전 폰 끄기는 알람을 밤 10시 반에 맞추고, 충전기를 침실 밖에 두세요.\n💪 응원: 작은 습관이 쌓이면 큰 변화가 됩니다. 이번 주도 화이팅!"))

if None in 달성률.values():
    print("⏳ A10-2 빈칸을 먼저 채우세요")
else:
    최고 = max(달성률, key=달성률.get)
    최저 = min(달성률, key=달성률.get)
    요약 = ", ".join(f"{k} {v}%" for k, v in 달성률.items())
    질문 = (f"너는 다정한 루틴 코치야. 내 일주일 습관 달성률은 {요약} 이야. "
            f"가장 잘한 습관은 {최고}, 가장 아쉬운 습관은 {최저}. "
            "'이번 주 잘한 점 / 다음 주 제안 2가지 / 응원 한마디' 형식으로 5줄 이내로 써 줘.")
    보고서 = AI(질문, 보여주기=True)
    with open("루틴보고서.txt", "w", encoding="utf-8") as f:
        f.write(f"[일주일 달성률] {요약}\n\n{보고서}\n")
    print("\n💾 루틴보고서.txt 저장 완료 — 왼쪽 📁 파일 탭에서 내려받기")
```

## 🏆 도전 문제 (선택)
**도전 — 최장 연속 기록**: 습관마다 O가 **가장 길게 이어진 날 수**를 계산해 보세요. (힌트: `연속`과 `최장` 두 변수를 두고, O면 연속 += 1, X면 연속 = 0)

```python
for 습관, 칸 in 기록.items():
    연속 = 최장 = 0
    for 표시 in 칸:
        연속 = 연속 + 1 if 표시 == "O" else 0
        최장 = max(최장, 연속)
    print(f"{습관:<12} 최장 {최장}일 연속 🔥")
```
