구글 시트로 다룬 **한빛시 도서관 예시 데이터**(수업용 가상 데이터)를 파이썬으로 다시 계산해서 **구글 시트 · AI 해석 · 파이썬** 세 결과를 교차 확인합니다.

````python
# 정답 확인 도우미 — 맨 먼저 한 번 실행하세요
def 확인(이름, 결과, 정답):
    if 결과 is None:
        print("⏳", 이름, ": 아직 비어 있어요")
    elif 결과 == 정답:
        print("✅", 이름, ": 정답입니다!", 결과)
    else:
        print("❌", 이름, ": 결과", 결과, "/ 기대", 정답)

print("준비 완료")
````

## 1. 데이터와 방문자 합계 · 평균
같은 번호끼리 한 달입니다. 예) `월[0]` = "1월", `방문자[0]` = 12400, `전자책[0]` = 2100

````python
월 = ["1월", "2월", "3월", "4월", "5월", "6월"]
방문자 = [12400, 11800, 13900, 14200, 15100, 9800]
전자책 = [2100, 2300, 2250, 2600, 2800, 3400]
비고 = ["", "", "", "", "", "6월 8~14일 시설 공사로 휴관"]

for i in range(len(월)):
    print(월[i], "| 방문자", 방문자[i], "| 전자책", 전자책[i], "|", 비고[i])

방문자_합계 = sum(방문자)
방문자_평균 = round(방문자_합계 / len(방문자), 1)
print()
print("방문자 합계:", 방문자_합계)
print("방문자 평균:", 방문자_평균)
확인("방문자 합계", 방문자_합계, 77200)
확인("방문자 평균", 방문자_평균, 12866.7)
````

## 2. ✏️ 전자책 합계
1번에서 방문자 합계를 구한 방법을 그대로 따라 하세요.

````python
전자책_합계 = sum(전자책)

print("전자책 합계:", 전자책_합계)
확인("전자책 합계", 전자책_합계, 15450)
````

**전자책 증가율**(1월 → 6월)도 확인해 봅니다. 증가율 = (나중 − 처음) ÷ 처음 × 100
`전자책[-1]`은 리스트의 **마지막** 값(6월)입니다.

````python
증가율 = round((전자책[-1] - 전자책[0]) / 전자책[0] * 100, 1)
print("전자책 1월 → 6월 증가율:", 증가율, "%")
확인("전자책 증가율(%)", 증가율, 61.9)
````

## 3. ✏️ 지난달보다 얼마나 변했나?
2월부터 시작해서 **이번 달 − 지난달**을 계산합니다. `i`번째 달의 지난달은 `i - 1`번째입니다.

````python
변화 = []   # 2월~6월의 방문자 변화를 차례로 담습니다

for i in range(1, len(방문자)):
    차이 = 방문자[i] - 방문자[i - 1]
    변화.append(차이)
    print(월[i], "방문자 변화:", 차이, 비고[i])
````

````python
확인("2월 변화", 변화[0], -600)
확인("6월 변화", 변화[-1], -5300)
````

## 4. 가장 크게 줄어든 달은?
`min(변화)`로 가장 작은(가장 많이 줄어든) 값을 찾고, 그 값이 몇 번째인지 `index()`로 찾습니다.
`변화[0]`은 2월이므로 **월 번호는 한 칸 뒤**(`+ 1`)입니다.

````python
if None in 변화:
    print("3번 빈칸을 먼저 채우세요")
else:
    가장_큰_감소 = min(변화)
    위치 = 변화.index(가장_큰_감소) + 1
    print("가장 크게 줄어든 달:", 월[위치], 가장_큰_감소, "명")
    print("그 달의 비고:", 비고[위치])
````

👉 6월에 방문자가 크게 줄었을 때 전자책은 늘었습니다. 그렇다고 **"전자책 때문에 방문자가 줄었다"** 고 말할 수 있을까요?
비고를 보면 **일주일 휴관**이 있었습니다. **함께 변했다고 원인은 아닙니다.**

## 5. 휴관을 반영해 하루 평균으로 비교
6월은 30일 중 7일 휴관 → **운영일 23일**, 5월은 31일 모두 운영했다고 가정합니다.

````python
오월_하루평균 = round(방문자[4] / 31, 1)
유월_하루평균 = round(방문자[5] / 23, 1)
print("5월 하루 평균:", 오월_하루평균, "명")
print("6월 하루 평균:", 유월_하루평균, "명")

한달_감소율 = round((방문자[5] - 방문자[4]) / 방문자[4] * 100, 1)
하루_감소율 = round((유월_하루평균 - 오월_하루평균) / 오월_하루평균 * 100, 1)
print()
print("한 달 합계로 보면", 한달_감소율, "%")
print("하루 평균으로 보면", 하루_감소율, "%")
````

# 🏆 도전 문제 (선택)
**파일 입출력 · csv.DictReader · 튜플 리스트 · 그래프**를 씁니다.

## 도전 A. CSV로 저장하고 딕셔너리로 읽기

````python
import csv

데이터 = """월,방문자,전자책,비고
1월,12400,2100,
2월,11800,2300,
3월,13900,2250,
4월,14200,2600,
5월,15100,2800,
6월,9800,3400,6월 8~14일 시설 공사로 휴관
"""
with open("도서관.csv", "w", encoding="utf-8") as f:
    f.write(데이터)

기록 = []
with open("도서관.csv", encoding="utf-8") as f:
    for 줄 in csv.DictReader(f):
        줄["방문자"] = int(줄["방문자"])
        줄["전자책"] = int(줄["전자책"])
        기록.append(줄)
print(기록[0])
print(기록[-1])
````

## 도전 B. ✏️ 전월 대비 변화를 튜플로 모으기
`(월, 방문자 변화, 전자책 변화, 비고)` 튜플을 `변화표`에 추가하세요.

````python
변화표 = []
for i in range(1, len(기록)):
    이번, 지난 = 기록[i], 기록[i - 1]
    변화표.append((이번["월"], 이번["방문자"] - 지난["방문자"], 이번["전자책"] - 지난["전자책"], 이번["비고"]))

for 이름, 방문, 전자, 메모 in 변화표:
    print(f"{이름:>3}  방문자 {방문:+7,}  전자책 {전자:+6,}  {메모}")
````

## 도전 C. 같은 방향? 반대 방향?
두 변화를 곱해서 **양수면 같은 방향**, **음수면 반대 방향**입니다.

````python
같은방향, 반대방향 = 0, 0
for 이름, 방문, 전자, 메모 in 변화표:
    if 방문 * 전자 > 0:
        같은방향 += 1
        판정 = "같은 방향"
    else:
        반대방향 += 1
        판정 = "반대 방향"
    print(이름, "방문자", "↑" if 방문 > 0 else "↓", "전자책", "↑" if 전자 > 0 else "↓", "→", 판정)

print("같은 방향", 같은방향, "개월 · 반대 방향", 반대방향, "개월")
if 변화표:
    확인("같은 방향 달 수", 같은방향, 2)
    확인("반대 방향 달 수", 반대방향, 3)
````

👉 반대 방향이 조금 많지만 **여섯 달 자료**로 규칙이라 말하기 어렵고, 가장 큰 감소는 **휴관** 때문입니다. (상관 ≠ 인과)
